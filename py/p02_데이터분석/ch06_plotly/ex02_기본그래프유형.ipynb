## 선 그래프(line plot)

```python
from datetime import datetime, timedelta

import pandas as pd
import numpy as np
import plotly.express as px
from plotly.subplots import make_subplots
```

```python
num = np.random.randn(30)
num
```

```python
# 시계열 데이터 생성
# 2024-01-01 부터 D(Daily) 하루씩 30일간
dates = pd.date_range('2026-06-01', periods=30, freq='D')
# np.random.randn(30): 평균 0, 표준편차 1에 따른 난수 30개
values = np.cumsum(np.random.randn(30)) + 100

time_series = pd.DataFrame({
    '날짜': dates,
    '주가': values
})
time_series
```

```python
# 1. 단일 선 그래프
fig = px.line(
    time_series, 
    x='날짜', 
    y='주가',
    title='일일 주가 변동',
    markers=True
)
fig
```

```python
# 2. 다중 선 그래프
# 데이터 생성
companies = ['삼성전자', 'LG전자', 'SK하이닉스']
stock_data = []

for company in companies:
    for i, date in enumerate(dates):
        price = 100 + np.cumsum(np.random.randn(1))[0] * (i + 1) / 10
        stock_data.append({
            '날짜': date,
            '회사': company,
            '주가': price
        })

# 데이터 분석
df_stocks = pd.DataFrame(stock_data)
df_stocks
```

## 히스토그램(histogram)

```python
# 1. 기본 히스토그램
# 시험 점수 분포 데이터
# 데이터 생성
# NumPy의 난수만 고정
np.random.seed(42)
# 평균 75, 표준편차 15
# np.random.normal(평균, 표준편차, 개수)
# 평균 75, 표준편차 15인 정규분포(normal distribution)에서 랜덤 숫자 1000개를 생성
scores = np.random.normal(75, 15, 1000)
# 0-100 사이로 제한
# np.clip(배열, 최소값, 최대값)
scores = np.clip(scores, 0, 100)
scores
```

```python
# 데이터 분석
df_hist = pd.DataFrame({'점수': scores})
df_hist
```

```python
# 데이터 시각화
fig = px.histogram(
    df_hist,
    x='점수',
    title='시험 점수 분포',
    nbins=20,  # 구간 수
    marginal='rug'  # 추가 통계 차트: rug, box, violin, histogram
)

'''
nbins=20: 데이터를 쪼갤 막대 격자(Bin)의 개수를 20개로 지정합니다. 0점부터 100점까지의 범위를 20개로 쪼갰으므로, 하나의 막대는 약 5점 단위(0~5, 5~10...)의 범위를 가지며 그 안에 속한 학생 수를 셉니다.
marginal='rug':히스토그램 상단(혹은 하단) 여백에 러그(Rug, 양탄자 술) 플롯을 추가합니다.데이터 포인트 1개당 아주 얇은 세로선 한 줄(|)을 그어주는 방식인데, 막대로 뭉뚱그려진 히스토그램 밑에 실제 데이터가 어디에 촘촘하게 몰려있고 어디가 텅 비었는지 미세하게 짚어주는 서브 차트 역할을 합니다.이 옵션에는 'rug' 외에도 박스플롯을 붙여주는 'box', 바이올린 플롯을 붙여주는 'violin' 등을 넣을 수 있습니다.
'''

fig.update_layout(
    xaxis_title='점수',
    yaxis_title='빈도'
)
fig
```

# 2. 그룹별 히스토그램

```python
# 성별 키 분포 비교
height_data = []
np.random.seed(42)

# 남성 키 데이터 (평균 175cm)
male_heights = np.random.normal(175, 6, 500)
for height in male_heights:
    height_data.append({'키': height, '성별': '남성'})

# 여성 키 데이터 (평균 162cm)
female_heights = np.random.normal(162, 5, 500)
for height in female_heights:
    height_data.append({'키': height, '성별': '여성'})

df_heights = pd.DataFrame(height_data)
df_heights
```

```python
# 데이터 시각화
fig = px.histogram(
    df_heights,
    x='키', 
    color='성별',
    title='성별 키 분포 비교',
    marginal='box',  # 박스플롯 추가
    opacity=0.7, # 불투명도
    barmode='overlay' # 배치 방식: overlay, group, stack, relative
)

fig.update_layout(
    xaxis_title='키 (cm)',
    yaxis_title='빈도'
)

fig
```

## 캔들스틱 차트(주가 데이터용)

```python
# 기본 캔들스틱 차트
# 주가 데이터 셍성
datetime.now()
```

```python
dates = [datetime.now() - timedelta(days=x) for x in range(30, 0, -1)]
dates
```

```python
np.random.seed(42)

ohlc_data = []
price = 100

for date in dates:
    open_price = price
    high_price = open_price + np.random.uniform(0, 5)
    low_price = open_price - np.random.uniform(0, 5)
    close_price = open_price + np.random.uniform(-3, 3)

    ohlc_data.append({
        '날짜': date,
        '시가': open_price,
        '고가': high_price,
        '저가': low_price,
        '종가': close_price
    })

    price = close_price

df_ohlc = pd.DataFrame(ohlc_data)
df_ohlc
```

```python
# 캔들스틱 차트 생성
import plotly.graph_objects as go

fig = go.Figure(data=go.Candlestick(
    x=df_ohlc['날짜'],
    open=df_ohlc['시가'],
    high=df_ohlc['고가'],
    low=df_ohlc['저가'],
    close=df_ohlc['종가'],
    name='주가'
))

fig.update_layout(
    title='일별 주가 캔들스틱 차트',
    xaxis_title='날짜',
    yaxis_title='가격',
    xaxis_rangeslider_visible=False # 하단 범위 슬라이더 제거
)

fig
```

```python
## 2. 거래량과 함께 표시
df_ohlc['거래량'] = np.random.randint(1000, 10000, len(df_ohlc))
```

```python
# 서브플롯 생성
fig = make_subplots(
    rows=2, 
    cols=1,
    shared_xaxes=True,
    vertical_spacing=0.03,
    subplot_titles=('주가', '거래량'),
    row_width=[0.2, 0.7]
)

fig
```

```python
# 캔들스틱 차트
fig.add_trace(go.Candlestick(
        x=df_ohlc['날짜'],
        open=df_ohlc['시가'],
        high=df_ohlc['고가'],
        low=df_ohlc['저가'],
        close=df_ohlc['종가'],
        name='주가'
    ), row=1, col=1)

fig.update_layout(
    title='주가와 거래량',
    xaxis_rangeslider_visible=False
)

fig
```

```python
# 거래량 박대 차트
fig.add_trace(
    go.Bar(
        x=df_ohlc['날짜'],
        y=df_ohlc['거래량'],
        name='거래량',
        marker_color='rgba(158,202,225,0.8)'
    ),
    row=2, col=1
)

fig
```
