```python
# 리스트(List)
lst = list(range(10))
print(lst)
print(lst * 2)

result = []
for num in lst:
    result.append(num * 2)

print(result)
```

```python
import numpy as np
# 배열(Array)
arr = np.arange(10)
print(arr)
print(arr * 2)
```

```python
import pandas as pd
import seaborn as sns

df = sns.load_dataset('tips')
df.head()
```

```python
df.info()
```

```python
# 결측치 개수 확인
df.isna().sum()
```

```python
# 주요 통계 지표(숫자 타입)
# 개수(count), 평균(mean), 표준편차(std), 최솟값(min), 분위수(25%, 50%, 75%), 최댓값(max)
df.describe()
```

```python
# 중볻값 확인
df.duplicated().sum()
```

```python
df[df.duplicated()]
```

```python
# 카테고리 타입의 값 확인
df['day'].unique()
```

```python
# smoker 유형에 따라 각 수치형 변수의 평균값 출력
# Pandas  2.0 버전 이후는 카테고리 유형이 자동 배제되지 않는다
df.groupby(['smoker']).mean(numeric_only=True)
```

```python
# 머신러닝 알고리즘은 모든 데이터가 수직형 데이터일 경우에만 학습 가능
# 원핫 인코딩 -> 카테고리형 데이터를 수치형 데이터로 변환
pd.get_dummies(df, dtype=int).head()
```
