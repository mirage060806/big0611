```python
# 모듈
# : 이란 함수나 변수 또는 클래스를 모아 놓은 파이썬 파일(*.py)이다.
# 모듈 만들기
# mod1.py
# def add(a, b):
#     return a + b

# def sub(a, b): 
#     return a-b
```

```python
# 모듈 불러오기
# import는 현재 디렉터리에 있는 파일이나 파이썬 라이브러리가 저장된 디렉터리에 있는 모듈만 불러올 수 있다.
# import 모듈_이름
import mod1

# mod1.py 파일에 있는 add 함수를 사용하려면 mod1.add처럼 모듈 이름 뒤에 도트 연산자(.)를 붙이고 함수 이름을 쓰면 된다.
# 모듈.함수
print(mod1.add(3, 4))
print(mod1.sub(4, 2))
```

```python
# 함수를 직접 import하면 모듈 이름을 붙이지 않고 바로 해당 모듈의 함수를 쓸 수 있다.
# from 모듈_이름 import 모듈_함수
from mod1 import add

# 함수
add(3, 4)
```
