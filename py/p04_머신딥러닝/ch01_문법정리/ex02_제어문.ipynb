## 2. 반복문

```python
# 데이터 유형: 숫자(정수), [리스트], '문자열'
# 변수: numbers, i
# 연산자: 대입(=), 비교(<, ==)
# 제어문: 조건문(if~elif~else), 반복문(for~in)
# 내장 함수: range(), print(), str()
# f-string: '{표현식} 문자열'

numbers = range(10)

for i in numbers:
    if i < 5:
        print(str(i) + '는 5보다 작습니다.')
    elif i == 5:
        print('{}는 5와 같습니다.'.format(i))
    else:
        print(f'{i}는 5보다 큽니다.')
```

```python
# 예시: 배터리가 10% 이상일 때만 데이터 수집 진행
battery = 15

while battery > 10:
    print("데이터를 수집 중입니다...")
    # battery = battery - 2  # 배터리 감소 표현
    battery -= 2  # 배터리 감소 표현
print("배터리가 부족하여 수집을 중단합니다.")
```

```python
# 리스트 컴프리헨션 (List Comprehension)
# 기존 방식: 0부터 9까지 숫자 중 짝수만 골라 리스트 만들기
even_numbers = []
for i in range(10):
    if i % 2 == 0:
        even_numbers.append(i)

# 한 줄로 줄이기 (결과는 동일)
even_numbers = [i for i in range(10) if i % 2 == 0]
print(even_numbers) # [0, 2, 4, 6, 8]
```

```python
# [참일 때 값 if 조건 else 거짓일 때 값 for 변수 in 반복]
result = [x if x % 2 == 0 else 0 for x in range(1, 6)]
print(result)
```

```python
# 중첩 반복문

# [표현식 for 변수 in 반복1 for 변수 in 반복2]
result = [x + y for x in [1, 2] for y in [10, 20]]
print(result)
```

```python
# 조건부 표현식(conditional expression)
# 참값 if 조건 else 거짓값
age = 20
result = "성인" if age >= 18 else "미성년자"
print(result)
```
