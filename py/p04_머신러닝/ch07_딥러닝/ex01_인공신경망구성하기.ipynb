```python
# 딥러닝
# 인공신경망 구성하기
# 유방암 데이터셋을 활용한 인공신경망 설계
""" 
문7-1> 유방암 데이터셋(Breast Cancer)을 활용한 인공신경망을 설계하시오.
- 가장 마지막 컬럼(target)은 종속변수(y), 나머지 컬럼은 모두 독립변수(x)로 활용
- 데이터 전처리는 고려하지 않음
- 은닉층은 1개로 설정하고 은닉층의 노드 수는 임의의 수로 설정할 것
"""
import pandas as pd
from sklearn.datasets import load_breast_cancer

# 데이터 불러오기
cancer = load_breast_cancer()
cancer_df = pd.DataFrame(data=cancer.data, columns=cancer.feature_names)
cancer_df['target'] = cancer.target
print(cancer_df.shape)
print(cancer_df['target'].unique())
```

```python
# 당뇨 데이터셋을 활용한 인공신경망 설계
""" 
문7-2> 당뇨 데이터셋(Diabetes)을 활용한 인공신경망을 설계하시오.
- 가장 마지막 컬럼(target)은 종속변수(y), 나머지 컬럼은 모두 독립변수(x)로 활용
- 데이터 전처리는 고려하지 않음
- 은닉층은 1개로 설정하고 은닉층의 노드 수는 임의의 수로 설정할 것
"""
import pandas as pd
from sklearn.datasets import load_diabetes

# 데이터 불러오기
diabetes = load_diabetes()
diabetes_df = pd.DataFrame(data=diabetes.data, columns=diabetes.feature_names)
diabetes_df['target'] = diabetes.target
print(diabetes_df.shape)
print(diabetes_df['target'].unique())
```
