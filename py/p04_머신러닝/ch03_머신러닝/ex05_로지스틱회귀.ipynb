```python
# 로지스틱 회귀(Logistic Regression)
# 로지스틱 회귀 기반 분류 모델 학습 및 예측
'''
# 문3-9> 주어진 코드를 사용하여 3개의 품종(Versicolor, Setosa, Virginica)을 분류하는 로지스틱 회귀 모델을 생성하고, 테스트셋을 사용하여 평가 지표를 확인하시오.
- 로지스틱 회귀 모델 생성 시, random_state=42로 설정
'''
```

```python
# 패키지 로딩
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression

# 데이터 로딩
iris = load_iris()

# 로딩한 데이터를 데이터프레임으로 생성
iris_df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
iris_df['label'] = iris.target

# 데이터 분할
X = iris_df.iloc[:, :4]
y = iris_df['label']

# 학습용 데이터와 테스트용 데이터로 나눔
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42, stratify=y)
```

```python
logistic_model = LogisticRegression(random_state=42)
logistic_model.fit(X_train, y_train)
print(logistic_model.score(X_test, y_test))
```
