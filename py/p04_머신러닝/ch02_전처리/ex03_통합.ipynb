```python
import pandas as pd
```

```python
# 두 개의 통합혈 데이터프레임
df1 = pd.DataFrame({
    'ID': ['1', '2', '3'],
    'Name': ['Alice', 'Bob', 'Charlie']
})

df2 = pd.DataFrame({
    'ID': ['4', '5', '6'],
    'Name': ['David', 'Eva', 'Frank']
})
```

```python
df1
```

```python
df2
```

```python
# 데이터프레임 df1과 df2를 행(Row) 기준으로 연결(위아래로 붙이기)
result_concat = pd.concat([df1, df2], axis=0, ignore_index=True)
print(result_concat)
```

```python
# ignore_index=True -> 기존 인덱스 무시
# axis 기본값은 0(위에서 아래로 붙이기)
pd.concat([df1, df2], axis=0, ignore_index=True)
```
