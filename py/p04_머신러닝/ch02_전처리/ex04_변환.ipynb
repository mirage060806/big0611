```python
# 데이터 변환
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.preprocessing import LabelEncoder, MinMaxScaler, StandardScaler
```

```python
data = {
    '이름': ['오리', '오리', '미금', '기흥', '미금', '기흥', '기흥'],
    '과목': ['수학', '과학', '수학', '과학', '과학', '수학', '수학'],
    '점수': [85, 90, 78, 88, 95, 80, 85]
}

df = pd.DataFrame(data)
df
```

```python
# 1. 데이터 집계(aggregation)
# 특정 열 기준 결합하여 집계 데이터 계산
# 과목별 평균 점수
df.groupby('과목')['점수'].mean()
```

```python
# 피벗 테이블 생성: pivot(), pivot_table(), crosstab()
# 피벗 테이블로 변환
df.pivot(index='이름', columns='과목', values='점수')
```

```python
# 피벗 테이블 생성 및 평균 계산
df.pivot_table(index='이름', columns='과목', values='점수', aggfunc='mean')
```

```python
# 이름과 과목에 따른 점수 합계
pd.crosstab(df['이름'], df['과목'], values=df['점수'], aggfunc='sum')
```

```python
# 2. 데이터 인코딩: 레이블 인코딩, one-hot 인코딩
# 예시 데이터
data = ['Red', 'Green', 'Blue', 'Green', 'Red']

# LabelEncoder 객체 생성
encoder = LabelEncoder()

# 'data'를 숫자형으로 변환
encoded_data = encoder.fit_transform(data)
print(encoded_data)
```

```python
data = {
    'Gender': ['Male', 'Female', 'Female', 'Male'],
    'City': ['Seoul', 'Busan', 'Incheon', 'Seoul']
}

df = pd.DataFrame(data)
df
```

```python
# 범주형 변수가 포함된 열에 대해 더미 변수를 생성
pd.get_dummies(df)
```

```python
# 특정 열(Gender)에 대해서만 더미 변수를 생성
pd.get_dummies(df, columns=['Gender'])
```

```python
# 범주의 카테고리에서 하나의 값을 제외하고 더미 변수를 생성
# 첫 번째 더미 변수 제거(drop_first=True) -> 사전순서대로 제거
pd.get_dummies(df, drop_first=True)
```

```python
# 3. 스케일 조정
# 샘플 데이터 생성
data = {
    '나이': [25, 30, 45, 50, 60],
    '연봉': [3500, 4000, 5000, 5500, 6500]
}

df = pd.DataFrame(data)
df
```

```python
# Min-Max 정규화
min_max_scaler = MinMaxScaler()
print('Min-Max 정규화된 데이터')
pd.DataFrame(min_max_scaler.fit_transform(df), columns=df.columns)
```

```python
# 표준화 수행
standard_scaler = StandardScaler()
print('표준화된 데이터')
pd.DataFrame(standard_scaler.fit_transform(df), columns=df.columns)
```

```python
# 사이킷런 패키지는 다양한 데이터셋을 제공한다.
# 붓꽃 데이터셋(iris data)
# - 꽃받침 길이와 너비, 꽃잎 길이와 너비의 4개의 특징 데이터에 따른 품종을 구분한 데이터로 총 150개의 값이 존재한다.
# 데이터 불러오기
iris = load_iris()
df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
df
```

```python
df['species'] = iris.target
```

```python
# 데이터 분포 확인
df.describe()

X = df.drop(columns=['species'])
```

```python
# 정규화 수행을 위한 객체 생성
minmax_scaler = MinMaxScaler()

# 데이터 변환을 위한 기준값 설정
minmax_scaler.fit(X)

# 정규화 수행
minmax_scaled_data = minmax_scaler.transform(X)

# 정규화 수행 결과를 포함하는 데이터프레임 생성
minmax_scaled_df = pd.DataFrame(data=minmax_scaled_data, columns=iris.feature_names)
minmax_scaled_df.describe()
```

```python
# 표준화 수행을 위한 객체 생성
standard_scaler = StandardScaler()

# 데이터 변환을 위한 기준값 설정
standard_scaler.fit(X)

# 표준화 수행
standard_scaled_data = standard_scaler.transform(X)

# 표준화 수행 결과를 포함하는 데이터프레임 생성
standard_scaled_df = pd.DataFrame(data=standard_scaled_data, columns=iris.feature_names)

# 통계 분포 확인
standard_scaled_df.describe()
```
