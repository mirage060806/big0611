```python
# K-means 알고리즘
# KMeans로 군집 모델 생성하기
# 군집 결과 시각화
# 군집 중심 시가화하기
'''
문4-4> 다음은 붓꽃 데이터셋에 대해 표준화를 수행한 결과(X_scaled)이다. X_scaled에 대해 KMeans 군집 방식으로 클러스터링을 수행하고, 예측값을 확인하시오.
- 클러스터링 수: 3개
'''

import sklearn

# 모델 다이어그램 출력을 텍스트 형식으로 변경
sklearn.set_config(display="text")

from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# 붓꽃 데이터셋 로드
data = load_iris()
X = data.data
y = data.target
```

```python
# 데이터 표준화 (스케일 조정)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# KMeans 모델 훈련
# 군집의 수를 3으로 설정
kmeans = KMeans(n_clusters=3, random_state=42)
kmeans.fit(X_scaled)
```

```python
# 군집화 결과 예측
y_pred = kmeans.predict(X_scaled)

print(y_pred)
```

```python
'''
문4-5> 문4-4에서 예측한 군집 결과와 원본 데이터에 대해서 시각화 하시오.
- 시각화 방법: 산점도(scatter plot)
- x축: X_scaled 4개의 특성 중 [0]값, y축: X_scaled 4개의 특성 중 [1]값
- 각 군집에 따라 다른 색상으로 표시
'''
# 군집화 결과 시각화 (산점도)
fig, ax = plt.subplots(ncols=2, figsize=(16, 4))

# 첫 번째 서브플롯: 군집화된 결과
sns.scatterplot(x=X_scaled[:, 0], y=X_scaled[:, 1], hue=y_pred, palette='Set1', ax=ax[0])
ax[0].set_title('KMeans Clustering Results')

# 두 번째 서브플롯: 실제 붓꽃 데이터의 클래스
sns.scatterplot(x=X_scaled[:, 0], y=X_scaled[:, 1], hue=y, palette='Set1', ax=ax[1])
ax[1].set_title('Actual Iris Classes')

plt.show()
```

```python
'''
문4-6> KMeans 클러스터링 결과 군집 중심을 출력하고, 문4-5의 KMeans 시각화에 군집 중심을 더하여 표시하시오.
'''
# 군집 중심 출력
print(kmeans.cluster_centers_)
```

```python
'''
[[ 0.57100359 -0.37176778  0.69111943  0.66315198]
 [-0.81623084  1.31895771 -1.28683379 -1.2197118 ]
 [-1.32765367 -0.373138   -1.13723572 -1.11486192]]
'''

# 군집 중심
cluster_centers = kmeans.cluster_centers_

sns.scatterplot(x=X_scaled[:, 0], y=X_scaled[:, 1], hue=y_pred, palette='Set1')

# 군집 중심 시각화
plt.scatter(cluster_centers[:, 0], cluster_centers[:, 1], marker='X', color='black')

plt.show()
```
