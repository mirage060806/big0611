```python
# 군집 알고리즘의 평가 방법
# KMeans로 군집 모델 생성하기
# 실루엣 점수 구하기
# 최적의 K값 구하기
'''
문4-7> 다음은 임의의 데이터셋에 대해서 KMeans 군집 분석(K=2)을 수행한 후, 실루엣 점수를 계산한 결과이다. 주어진 데이터셋으로 KMeans 군집 분석할 때 K값을 2~5의 값으로 설정하고자 할 때, 최적의 K값은 무엇인지 구하시오.
'''
import numpy as np
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs
from sklearn.metrics import silhouette_score
```

```python
# 간단한 데이터셋 생성 (두 개의 특성, 3개의 군집)
X, y = make_blobs(n_samples=300, centers=3, random_state=42)

kmeans = KMeans(n_clusters=2, random_state=42)
kmeans.fit(X)

# 군집 예측 결과
y_pred = kmeans.predict(X)
```

```python
# 실루엣 점수 계산
score = silhouette_score(X, y_pred)
print(f'K=2, Silhouette Score: {score:.3f}') # K=2, Silhouette Score: 0.705
```

```python
# 실루엣 점수 기록을 위한 리스트
silhouette_scores = []

# K 값을 3부터 5까지 변경하여 KMeans 모델 훈련
for k in range(3, 6):
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X)

    # 군집 예측 결과
    y_pred = kmeans.predict(X)

    # 실루엣 점수 계산
    sil_score = silhouette_score(X, y_pred)
    silhouette_scores.append(sil_score)

    # 결과 출력
    print(f'K={k}, Silhouette Score: {sil_score:.3f}')
```

```python
# 가장 좋은 K값 찾기 (최고의 실루엣 점수)
# 3부터 시작했으므로 인덱스에 3을 더함
best_k = np.argmax(silhouette_scores) + 3
print(f'Best K: {best_k}')
```
